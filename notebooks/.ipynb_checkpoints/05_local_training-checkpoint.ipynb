# 05 - Local training (laptop / workstation)

The local counterpart of `04_colab_training.ipynb`. Same data, same models, same
honest reporting - but it runs **on this machine** instead of in Colab.

| | 04 (Colab) | 05 (this notebook) |
|---|---|---|
| Where it runs | Colab only, hard guard | anywhere TensorFlow is importable |
| Data source | uploaded `data/final/` on Drive | the repo's own `data/final/` |
| Model output | Drive `trained/` | the repo's `models/` |
| Runtime | free GPU, minutes | your CPU, likely much longer |
| Defaults | full data, 60 epochs | tail-subsampled, 25 epochs, early stop |

> ### Read this before you start
>
> **No accuracy, profitability or trading-edge claim is made by this notebook.**
> It prints measurements, and it prints the naive baselines next to them so the
> measurements mean something.
>
> **Results from this notebook are not the results this project documents.**
> It defaults to a chronological *tail subset* of the training split so that a
> first run finishes while you are still watching it. A subset run answers
> "does the pipeline work and roughly what happens" - it is **not** a result to
> put in a report. Set `SUBSET_TRAIN = None` for the full training split.
>
> **A local run will not reproduce a Colab run exactly.** Different device,
> different kernel selection, different reduction order - even with every seed
> fixed. Treat small differences between the two as noise.
>
> 04 remains the reference notebook. This one exists for fast iteration: you can
> edit a feature in `src/features.py`, re-run `pipeline.py`, and immediately try
> the effect, without a round trip through Drive.

## 2. Runtime configuration

Everything you are likely to change lives here.

| knob | what it does |
|---|---|
| `SMOKE` | 1 epoch on a tiny slice. Plumbing check only - the numbers are meaningless. |
| `SUBSET_TRAIN` | use only the most recent N training samples. `None` = all 218,515. |
| `EPOCHS` | upper bound; early stopping usually fires first. |
| `BATCH` | larger is faster on CPU. Lower it if you run out of memory. |
| `UNITS` | recurrent width. 96 matches notebook 04. |
| `RUN_XGBOOST` / `RUN_LSTM` | turn the optional comparison models on or off. |
| `DETERMINISTIC` | bit-reproducible ops at a real speed cost. Off by default. |

`SUBSET_TRAIN` takes the **most recent** N samples, not a random N. The subset
stays contiguous and chronologically adjacent to the validation split, so the
data immediately before validation is never thrown away. It is still strictly
before the purge boundary - the test split is never involved in any choice here.


````python
SEED = 42
DETERMINISTIC = False          # True -> tf.config.experimental.enable_op_determinism()

# ---- runtime budget -------------------------------------------------------
SMOKE = False                  # True -> 1 epoch on a tiny slice (plumbing check)
SUBSET_TRAIN = 60_000          # None -> use the full 218,515-sample training split
EPOCHS = 25
BATCH = 1024
UNITS = 64                     # 96 matches notebook 04; 64 is kinder to a CPU
DROPOUT = 0.25
LEARNING_RATE = 1e-3
PATIENCE = 5                   # early-stopping epochs
RUN_XGBOOST = True
RUN_LSTM = False               # flip to True for the full 04 comparison
XGB_ESTIMATORS = 300

os.environ["PYTHONHASHSEED"] = str(SEED)
random.seed(SEED)
np.random.seed(SEED)
keras.utils.set_random_seed(SEED)
if DETERMINISTIC:
    tf.config.experimental.enable_op_determinism()

# ---- threads ---------------------------------------------------------------
# Leave a core or two free so the machine stays usable during a long run.
N_THREADS = max(1, (os.cpu_count() or 4) - 2)
tf.config.threading.set_intra_op_parallelism_threads(N_THREADS)
tf.config.threading.set_inter_op_parallelism_threads(max(1, N_THREADS // 2))
print(f"threads: {N_THREADS} intra-op, {max(1, N_THREADS // 2)} inter-op")

# ---- paths -----------------------------------------------------------------
def find_project_root() -> Path:
    """Locate the repo whether this notebook runs from notebooks/ or the root."""
    for cand in (Path.cwd(), Path.cwd().parent, Path.cwd().parent.parent):
        if (cand / "data" / "final" / "dataset_metadata.json").exists():
            return cand
    raise FileNotFoundError(
        "Could not find data/final/dataset_metadata.json. Run "
        "`python src/pipeline.py` first, and start Jupyter from the project "
        "root or from notebooks/."
    )


ROOT = find_project_root()
DATA_DIR = ROOT / "data" / "final"
OUT_DIR = ROOT / "models"
OUT_DIR.mkdir(parents=True, exist_ok=True)

sys.path.insert(0, str(ROOT / "src"))          # reuse the pipeline's own code

print("project root:", ROOT)
print("data dir    :", DATA_DIR)
print("output dir  :", OUT_DIR)
if SMOKE:
    print("\n*** SMOKE MODE - numbers from this run mean nothing ***")
````

## 3. Load the prepared dataset

Produced by `python src/pipeline.py`. See section 13 of `PROJECT_USAGE.txt` for
the full file list.

**The one thing to internalise:** `X_*.npy` is a *contiguous 2D block*, and
`starts_*.npy` says where each sample's window begins inside it. Gap-excluded
windows leave holes in the offset sequence, so

```python
window_k = X_block[starts[k] : starts[k] + LOOKBACK]      # correct
window_k = X_block[k : k + LOOKBACK]                      # WRONG - splices gaps
```

`y_*.npy`, `y_dir_*.npy` and `timestamps_*.npy` already hold exactly one row per
sample - no `[LOOKBACK-1:]` offset anywhere.


````python
REQUIRED = [
    "X_train.npy", "X_val.npy", "X_test.npy",
    "starts_train.npy", "starts_val.npy", "starts_test.npy",
    "y_train.npy", "y_val.npy", "y_test.npy",
    "y_dir_train.npy", "y_dir_val.npy", "y_dir_test.npy",
    "timestamps_train.npy", "timestamps_val.npy", "timestamps_test.npy",
    "scaler.pkl", "target_scaler.pkl", "dataset_metadata.json",
]
missing = [f for f in REQUIRED if not (DATA_DIR / f).exists()]
if missing:
    raise FileNotFoundError(
        "Missing files in " + str(DATA_DIR) + ": " + ", ".join(missing) +
        "\nRun `python src/pipeline.py` to regenerate them."
    )

X_train = np.load(DATA_DIR / "X_train.npy")
X_val = np.load(DATA_DIR / "X_val.npy")
X_test = np.load(DATA_DIR / "X_test.npy")

s_train = np.load(DATA_DIR / "starts_train.npy")
s_val = np.load(DATA_DIR / "starts_val.npy")
s_test = np.load(DATA_DIR / "starts_test.npy")

y_train = np.load(DATA_DIR / "y_train.npy")
y_val = np.load(DATA_DIR / "y_val.npy")
y_test = np.load(DATA_DIR / "y_test.npy")

d_train = np.load(DATA_DIR / "y_dir_train.npy").astype(np.int64)
d_val = np.load(DATA_DIR / "y_dir_val.npy").astype(np.int64)
d_test = np.load(DATA_DIR / "y_dir_test.npy").astype(np.int64)

ts_train = np.load(DATA_DIR / "timestamps_train.npy")     # int64 epoch ms, UTC
ts_val = np.load(DATA_DIR / "timestamps_val.npy")
ts_test = np.load(DATA_DIR / "timestamps_test.npy")

with open(DATA_DIR / "scaler.pkl", "rb") as fh:
    scaler = pickle.load(fh)
with open(DATA_DIR / "target_scaler.pkl", "rb") as fh:
    target_scaler = pickle.load(fh)
with open(DATA_DIR / "dataset_metadata.json", encoding="utf-8") as fh:
    META = json.load(fh)

LOOKBACK = META["lookback"]
HORIZONS = list(META["horizons"].keys())                  # LABELS: "15m".."90m"
N_H = len(HORIZONS)
N_F = X_train.shape[1]

# The horizon LABELS are strings, so max(HORIZONS) would be "90m" - not 6.
# Anything arithmetic (purge clearance, label reach) needs the candle counts.
HORIZON_CANDLES = [int(META["horizons"][h]["candles"]) for h in HORIZONS]
HORIZON_MAX = max(HORIZON_CANDLES)                        # 6 candles = 90m
print(f"horizon labels  : {HORIZONS}")
print(f"horizon candles : {HORIZON_CANDLES}  -> HORIZON_MAX = {HORIZON_MAX}")

# ---- structural checks: misaligned blocks/offsets would be silently fatal ----
for name, block, starts, y, d in (("train", X_train, s_train, y_train, d_train),
                                  ("val", X_val, s_val, y_val, d_val),
                                  ("test", X_test, s_test, y_test, d_test)):
    assert len(starts) == len(y) == len(d), f"{name}: offsets/targets misaligned"
    assert starts.min() >= 0, f"{name}: negative window offset"
    assert starts.max() + LOOKBACK <= len(block), f"{name}: window past block end"
    assert y.shape[1] == N_H and d.shape[1] == N_H, f"{name}: wrong horizon count"
    assert np.isfinite(block).all(), f"{name}: block contains NaN/Inf"

print(f"lookback={LOOKBACK}  features={N_F}  horizons={HORIZONS}")
print(f"\n{'split':6} {'block rows':>12} {'samples':>10}   target period")
for name, block, starts, ts in (("train", X_train, s_train, ts_train),
                                ("val", X_val, s_val, ts_val),
                                ("test", X_test, s_test, ts_test)):
    t0 = pd.to_datetime(int(ts[0]), unit="ms", utc=True)
    t1 = pd.to_datetime(int(ts[-1]), unit="ms", utc=True)
    print(f"{name:6} {len(block):>12,} {len(starts):>10,}   {t0:%Y-%m-%d} -> {t1:%Y-%m-%d}")

n_holes = int((np.diff(s_train) != 1).sum())
print(f"\ntrain offsets contain {n_holes} hole(s) - gap-excluded windows.")
print("That is exactly why windows must be read via starts_*.npy.")
````

## 4. Batching

`SequenceBatcher` in `src/sequences.py` does the windowing: it reads windows
from the 2D block through the saved offsets and materialises one batch at a
time, so a `(218515, 64, 49)` float32 tensor (3.9 GB) is never allocated.

`WindowSequence` below wraps it in `keras.utils.PyDataset` - the supported Keras 3
way to stream a custom Python sequence - and adds the direction head's targets.
The windowing logic itself is *the project's own code*, not a reimplementation,
so training and inference cannot drift apart.


````python
CANDLE_MS = 15 * 60 * 1000          # the 15m grid, in milliseconds


def chronology_guard(ts_train, ts_val, lookback, horizon_max):
    """
    Re-derive the purge invariant here rather than assuming it still holds.

    WHY THIS USES TIMESTAMPS, NOT starts_*.npy:
      starts_*.npy hold BLOCK-RELATIVE offsets - starts_val[0] is 0, because
      each split's offsets are measured from the start of its own block. They
      are correct for windowing and meaningless for comparing two splits.
      timestamps_*.npy hold each sample's target time (the timestamp of the
      LAST candle in its window) in absolute UTC, so they ARE comparable.

    THE INVARIANT:
      Let the last train window occupy rows [a, a+63] and the first validation
      window [b, b+63]. The last train label is resolved from close[t] and
      close[t+h] with t = a+63, so it reads as far as row a+63+horizon_max.
      Nothing in it may touch the validation features at rows b..b+63, so
          b - a  >=  lookback + horizon_max
      Both window spans are shifted by the same 63 rows, so
          b - a  ==  (ts_val[0] - ts_train[-1]) in candles
      which is what makes this checkable from the shipped timestamps alone.

    Returns the clearance in candles: how many rows of margin are left over.
    """
    gap_candles = (int(ts_val[0]) - int(ts_train[-1])) // CANDLE_MS
    required = lookback + horizon_max
    if gap_candles < required:
        raise AssertionError(
            f"Training labels overlap the validation feature window: the splits "
            f"are {gap_candles} candles apart, but lookback {lookback} + max "
            f"horizon {horizon_max} = {required} are needed. The purge buffer "
            f"is too small - re-run src/pipeline.py and its audit."
        )
    return gap_candles - required


# The windowing itself is the project's OWN code, reused rather than reimplemented,
# so training and inference cannot drift apart.
from sequences import SequenceBatcher
print("SequenceBatcher loaded from", Path(sys.modules["sequences"].__file__).name)


class WindowSequence(keras.utils.PyDataset):
    """
    Streams (X_batch, {return_head, direction_head}) windows from a 2D block.

    Window k is X_block[starts[k] : starts[k] + lookback] - never
    X_block[k : k + lookback], which would splice across the holes left by
    gap-excluded windows.

    `shuffle=True` permutes the order in which SAMPLES are visited, once per
    epoch. It never mixes splits and never reorders time inside a window.
    """

    def __init__(self, X_block, y_scaled, y_dir, starts, *, lookback,
                 batch_size=512, shuffle=False, seed=None, **kwargs):
        super().__init__(**kwargs)
        self.core = SequenceBatcher(X_block, y_scaled, starts, lookback=lookback,
                                    batch_size=batch_size, shuffle=False)
        self.ydir = np.asarray(y_dir, dtype=np.int64)
        self.n = int(self.core.n_samples)
        self.shuffle = bool(shuffle)
        self.rng = np.random.default_rng(seed)
        self.order = np.arange(self.n, dtype=np.int64)

    def __len__(self):
        return int(np.ceil(self.n / self.core.batch_size))

    def __getitem__(self, i):
        chunk = self.order[i * self.core.batch_size:(i + 1) * self.core.batch_size]
        Xb, yb = self.core.materialise(chunk)
        return Xb, {"return_head": yb, "direction_head": self.ydir[chunk]}

    def on_epoch_end(self):
        if self.shuffle:
            self.order = self.rng.permutation(self.n)


from sequences import SequenceBatcher           # the project's own windowing code

clearance = chronology_guard(ts_train, ts_val, LOOKBACK, HORIZON_MAX)
print(f"purge clearance: {clearance} candle(s) of margin between the last "
      f"train label and the first validation feature candle")
print("(the purge is sized to be exactly sufficient, so a small clearance is "
      "expected and correct - audit check 11 measures the same thing)")

# ---- optional chronological tail subset of the TRAINING split only ---------
X_fit, y_fit, d_fit, s_fit = X_train, y_train, d_train, s_train
if SMOKE:
    n_keep = min(4_000, len(y_train))
elif SUBSET_TRAIN is not None:
    n_keep = min(int(SUBSET_TRAIN), len(y_train))
else:
    n_keep = len(y_train)

if n_keep < len(y_train):
    cut = len(y_train) - n_keep
    s_fit = s_train[cut:]
    y_fit = y_train[cut:]
    d_fit = d_train[cut:]
    lo = int(s_fit[0])
    X_fit = X_train[lo:]                     # keep the block contiguous
    s_fit = (s_fit - lo).astype(np.int64)
    print(f"using the most recent {n_keep:,} of {len(y_train):,} training samples "
          f"(block trimmed to {len(X_fit):,} rows)")

# NOTE: y_fit stays RAW here; the scaled copy below is what the network sees.
MEAN_T = target_scaler.mean_.astype(np.float32)
SCALE_T = target_scaler.scale_.astype(np.float32)


def to_scaled(y):
    return ((y - MEAN_T) / SCALE_T).astype(np.float32)


def from_scaled(y):
    return (y * SCALE_T + MEAN_T).astype(np.float32)


fit_seq = WindowSequence(X_fit, to_scaled(y_fit), d_fit, s_fit, lookback=LOOKBACK,
                         batch_size=BATCH, shuffle=True, seed=SEED)
val_seq = WindowSequence(X_val, to_scaled(y_val), d_val, s_val, lookback=LOOKBACK,
                         batch_size=BATCH, shuffle=False)
test_seq = WindowSequence(X_test, to_scaled(y_test), d_test, s_test, lookback=LOOKBACK,
                          batch_size=BATCH, shuffle=False)

Xb, yb = fit_seq[0]
print("\none batch  :", Xb.shape, {k: v.shape for k, v in yb.items()})
print("batches    :", len(fit_seq), "fit |", len(val_seq), "val |", len(test_seq), "test")

# A shuffled batch must still be 64 contiguous candles per row.
assert Xb.shape[1] == LOOKBACK
first_after_shuffle = int(fit_seq.core.starts[fit_seq.order[0]])
assert first_after_shuffle + LOOKBACK <= len(X_fit)
print("shuffled batch windows start inside the block: OK")
````

## 5. Evaluation utilities and the naive baselines

Reported **per horizon**, never as a single pooled number: a model that is useful
at 15m and useless at 90m has to show up that way.

Two floors any real model must clear:

* **regression** - predict a zero return. Its MAE *is* the mean absolute return,
  so it grows with horizon volatility.
* **direction** - always predict the majority training class. On this test set
  that class is NEUTRAL, worth roughly 54% accuracy while being worthless. This
  is why macro-F1, not accuracy, is the headline number here.


````python
from sklearn.metrics import (accuracy_score, confusion_matrix, f1_score,
                             precision_recall_fscore_support)

CLASS_NAMES = ["DOWN", "NEUTRAL", "UP"]


def _safe_corr(a, b):
    """Pearson r, or NaN when either side is constant.

    The naive zero-return baseline predicts a constant, and the correlation of
    anything with a constant is undefined. Returning NaN keeps that honest;
    letting np.corrcoef divide by a zero std just emits a warning and yields
    NaN anyway, noisily.
    """
    if np.std(a) == 0 or np.std(b) == 0:
        return float("nan")
    return float(np.corrcoef(a, b)[0, 1])


def regression_metrics(y_true, y_pred, horizons=HORIZONS):
    rows = {}
    for i, h in enumerate(horizons):
        e = y_pred[:, i] - y_true[:, i]
        rows[h] = {
            "MAE_%": float(np.mean(np.abs(e))),
            "RMSE_%": float(np.sqrt(np.mean(e ** 2))),
            "bias_%": float(np.mean(e)),
            "corr": _safe_corr(y_true[:, i], y_pred[:, i]),
        }
    return pd.DataFrame(rows).T


def direction_metrics(y_true_dir, y_pred_dir, horizons=HORIZONS):
    """y_*_dir are (n, 6) with values in {0,1,2}."""
    rows = {}
    for i, h in enumerate(horizons):
        t, p = y_true_dir[:, i], y_pred_dir[:, i]
        prec, rec, f1, sup = precision_recall_fscore_support(
            t, p, labels=[0, 1, 2], zero_division=0)
        rows[h] = {
            "accuracy": accuracy_score(t, p),
            "macro_F1": f1_score(t, p, average="macro", zero_division=0),
            "DOWN_prec": prec[0], "DOWN_rec": rec[0], "DOWN_f1": f1[0],
            "NEUTRAL_prec": prec[1], "NEUTRAL_rec": rec[1], "NEUTRAL_f1": f1[1],
            "UP_prec": prec[2], "UP_rec": rec[2], "UP_f1": f1[2],
            "n": int(len(t)),
        }
    return pd.DataFrame(rows).T


def show_confusion(y_true_dir, y_pred_dir, horizon_index=0):
    cm = confusion_matrix(y_true_dir[:, horizon_index], y_pred_dir[:, horizon_index],
                          labels=[0, 1, 2])
    return pd.DataFrame(cm, index=[f"true_{c}" for c in CLASS_NAMES],
                        columns=[f"pred_{c}" for c in CLASS_NAMES])


def majority_class_baseline(y_true_dir):
    """Always predict the most frequent TRAINING class, per horizon."""
    preds = np.zeros_like(y_true_dir)
    for i in range(y_true_dir.shape[1]):
        vals, counts = np.unique(d_train[:, i], return_counts=True)
        preds[:, i] = vals[np.argmax(counts)]
    return preds


base_reg = regression_metrics(y_test, np.zeros_like(y_test))
print("NAIVE REGRESSION (predict 0% return)")
display(base_reg)

base_dir_pred = majority_class_baseline(d_test)
base_dir = direction_metrics(d_test, base_dir_pred)
print("\nNAIVE DIRECTION (majority train class per horizon)")
display(base_dir)
````

## 6. Model 1 - XGBoost

A tree model cannot consume a sequence, so each 64-candle window is compressed
into a fixed-length vector: the last candle plus the mean, std, min and max of
every feature across the window (5 x 49 = 245 inputs).

This is a genuinely strong tabular baseline. Do not assume the recurrent models
will beat it - reporting it honestly is the point of including it.

Direction comes from the **frozen training thresholds** in the metadata, not from
a threshold re-derived here. Re-deriving on validation or test would be leakage.


````python
if RUN_XGBOOST:
    import xgboost as xgb

    def aggregate_windows(X_block, starts, lookback):
        """
        Block + offsets -> (n_samples, 5F): last candle, plus mean, std, min and
        max over each window.

        Indexing goes through `starts`, never X_block[k : k + lookback], because
        gap-excluded windows leave holes in the offset sequence.
        """
        starts = np.asarray(starts, dtype=np.int64)
        rows = starts[:, None] + np.arange(lookback, dtype=np.int64)[None, :]
        w = X_block[rows]                              # (n_samples, lookback, F)
        return np.concatenate(
            [w[:, -1, :], w.mean(axis=1), w.std(axis=1), w.min(axis=1), w.max(axis=1)],
            axis=1,
        ).astype(np.float32)

    t0 = time.time()
    A_fit = aggregate_windows(X_fit, s_fit, LOOKBACK)
    A_val = aggregate_windows(X_val, s_val, LOOKBACK)
    A_test = aggregate_windows(X_test, s_test, LOOKBACK)
    assert len(A_fit) == len(y_fit), "aggregated windows and targets disagree"
    print("aggregated:", A_fit.shape, A_val.shape, A_test.shape,
          f"in {time.time() - t0:.1f}s")

    THR = {h: (META["thresholds"]["per_horizon"][h]["lower_percent"],
               META["thresholds"]["per_horizon"][h]["upper_percent"])
           for h in HORIZONS}
    print("thresholds loaded from metadata (frozen, train-derived):")
    for h in HORIZONS:
        print(f"    {h:>4}: [{THR[h][0]:+.4f}%, {THR[h][1]:+.4f}%]")

    xgb_val_pred, xgb_test_pred = {}, {}
    xgb_val_dir, xgb_test_dir = {}, {}

    for i, h in enumerate(HORIZONS):
        t0 = time.time()
        model = xgb.XGBRegressor(
            n_estimators=XGB_ESTIMATORS, max_depth=6, learning_rate=0.05,
            subsample=0.8, colsample_bytree=0.8, reg_lambda=1.0,
            early_stopping_rounds=30, random_state=SEED, n_jobs=N_THREADS,
            tree_method="hist",
        )
        model.fit(A_fit, y_fit[:, i], eval_set=[(A_val, y_val[:, i])], verbose=False)
        xgb_val_pred[i] = model.predict(A_val)
        xgb_test_pred[i] = model.predict(A_test)

        lo, hi = THR[h]
        xgb_val_dir[i] = np.digitize(xgb_val_pred[i], [lo, hi])
        xgb_test_dir[i] = np.digitize(xgb_test_pred[i], [lo, hi])
        print(f"  {h:>4}: {model.best_iteration + 1:>4} trees, {time.time() - t0:5.1f}s")

    xgb_val_pred = np.column_stack([xgb_val_pred[i] for i in range(N_H)])
    xgb_test_pred = np.column_stack([xgb_test_pred[i] for i in range(N_H)])
    xgb_val_dir = np.column_stack([xgb_val_dir[i] for i in range(N_H)])
    xgb_test_dir = np.column_stack([xgb_test_dir[i] for i in range(N_H)])

    print("\nXGBoost - VALIDATION")
    display(regression_metrics(y_val, xgb_val_pred))
    display(direction_metrics(d_val, xgb_val_dir))
else:
    print("RUN_XGBOOST = False - skipped")
````

## 7. Models 2 and 3 - GRU and LSTM (multi-task)

Shared recurrent trunk, two heads:

* `return_head` - linear, 6 outputs, MSE against **scaled** returns
* `direction_head` - softmax, 6 x 3 outputs, sparse categorical cross-entropy

The multi-task setup is the point. The direction head pushes the trunk toward
representations that separate the three classes; the regression head keeps the
magnitude information the class labels throw away. Training only the direction
head discards how big the move was; training only the return head gives a model
that regresses to the mean, which is precisely what the naive baseline does.

**Regressing on scaled returns, not raw percent.** Returns are ~0.39% at 15m, so
raw MSE gradients are tiny. Predictions are inverted back to percent before any
metric is computed.


````python
def build_multi_task_model(cell_type="GRU", units=UNITS, dropout=DROPOUT,
                          learning_rate=LEARNING_RATE, n_h=N_H, n_f=N_F,
                          lookback=LOOKBACK):
    layer = keras.layers.GRU if cell_type.upper() == "GRU" else keras.layers.LSTM

    inputs = keras.Input(shape=(lookback, n_f), name="window")
    x = layer(units, return_sequences=True, name=f"{cell_type.lower()}_1")(inputs)
    x = keras.layers.Dropout(dropout, name="drop_1")(x)
    x = layer(units, return_sequences=False, name=f"{cell_type.lower()}_2")(x)
    x = keras.layers.Dropout(dropout, name="drop_2")(x)

    return_head = keras.layers.Dense(n_h, name="return_head")(x)

    d = keras.layers.Dense(n_h * 3, activation="softmax", name="direction_logits")(x)
    direction_head = keras.layers.Reshape((n_h, 3), name="direction_head")(d)

    model = keras.Model(
        inputs=inputs,
        outputs={"return_head": return_head, "direction_head": direction_head},
        name=f"{cell_type.lower()}_multitask",
    )
    model.compile(
        optimizer=keras.optimizers.Adam(learning_rate),
        loss={"return_head": "mse",
              "direction_head": keras.losses.SparseCategoricalCrossentropy()},
        loss_weights={"return_head": 1.0, "direction_head": 1.0},
        metrics={"return_head": ["mae"],
                 "direction_head": ["sparse_categorical_accuracy"]},
    )
    return model


# Build one and push a dummy batch through it, so a shape mistake surfaces here
# rather than 20 minutes into training.
_probe = build_multi_task_model("GRU")
_probe.summary()
_out = _probe.predict(np.zeros((3, LOOKBACK, N_F), dtype=np.float32), verbose=0)
print("\nshape check:", {k: v.shape for k, v in _out.items()})
expected = {"return_head": (3, N_H), "direction_head": (3, N_H, 3)}
assert {k: v.shape for k, v in _out.items()} == expected, "head shapes are wrong"
print("head shapes OK:", expected)
del _probe, _out
````

````python
def train_recurrent(cell_type, epochs=EPOCHS, units=UNITS, dropout=DROPOUT,
                    lr=LEARNING_RATE, batch_size=BATCH, tag=None):
    """Fit one multi-task model with early stopping on the validation split."""
    tag = tag or cell_type.lower()
    keras.utils.set_random_seed(SEED)          # same init for every architecture
    model = build_multi_task_model(cell_type, units=units, dropout=dropout,
                                   learning_rate=lr)

    ckpt = OUT_DIR / f"{tag}_best.keras"
    callbacks = [
        keras.callbacks.EarlyStopping(monitor="val_loss", patience=PATIENCE,
                                      restore_best_weights=True, verbose=1),
        keras.callbacks.ReduceLROnPlateau(monitor="val_loss", factor=0.5,
                                          patience=2, min_lr=1e-5, verbose=1),
        keras.callbacks.ModelCheckpoint(ckpt, monitor="val_loss",
                                        save_best_only=True, verbose=0),
    ]

    t0 = time.time()
    history = model.fit(fit_seq, validation_data=val_seq,
                        epochs=epochs, callbacks=callbacks, verbose=2)
    elapsed = time.time() - t0

    n_epochs = len(history.history["loss"])
    print(f"\n{cell_type}: {n_epochs} epoch(s) in {elapsed / 60:.1f} min "
          f"({elapsed / max(1, n_epochs):.0f}s/epoch)")
    print(f"best weights: {ckpt}")
    return model, history, elapsed


def predict_recurrent(model, seq):
    raw = model.predict(seq, verbose=0)
    returns = from_scaled(raw["return_head"])
    direction = raw["direction_head"].argmax(axis=-1).astype(np.int64)
    return returns, direction
````

## 8. Train the GRU (primary architecture)

Watch the seconds-per-epoch line that `train_recurrent` prints. Multiply it by
`EPOCHS` to see the worst-case budget before committing to the LSTM as well.


````python
gru_model, gru_hist, gru_time = train_recurrent("GRU", tag="gru")

gru_val_pred, gru_val_dir = predict_recurrent(gru_model, val_seq)

print("=== GRU - VALIDATION REGRESSION ===")
display(regression_metrics(y_val, gru_val_pred))
print("=== GRU - VALIDATION DIRECTION ===")
display(direction_metrics(d_val, gru_val_dir))
````

````python
if RUN_LSTM:
    lstm_model, lstm_hist, lstm_time = train_recurrent("LSTM", tag="lstm")
    lstm_val_pred, lstm_val_dir = predict_recurrent(lstm_model, val_seq)
    print("=== LSTM - VALIDATION REGRESSION ===")
    display(regression_metrics(y_val, lstm_val_pred))
    print("=== LSTM - VALIDATION DIRECTION ===")
    display(direction_metrics(d_val, lstm_val_dir))
else:
    print("RUN_LSTM = False - skipped. Set it to True in section 2 for the full")
    print("comparison that notebook 04 produces (roughly doubles training time).")
````

## 9. Choose the architecture on validation - never on test

Validation macro-F1 decides. Test is then touched **once**, by the winner only.

If you find yourself looking at the test numbers, changing something, and
looking again, you no longer have a test set - you have a second validation set
with extra steps, and every number after that is optimistic.


````python
candidates = {"GRU": direction_metrics(d_val, gru_val_dir)["macro_F1"].mean()}
if RUN_LSTM:
    candidates["LSTM"] = direction_metrics(d_val, lstm_val_dir)["macro_F1"].mean()

print("mean validation macro-F1:")
for k, v in candidates.items():
    print(f"  {k:5} {v:.4f}")

BEST = max(candidates, key=candidates.get)
print("\nselected on validation:", BEST)
print("A small difference is a hypothesis, not a conclusion - re-run with "
      "different seeds before reading anything into it.")
````

## 12. Backtest scaffold (deliberately empty)

**No backtest is produced or claimed here.** Directional accuracy is not
profitability. What is missing before any P&L statement would mean anything:

* entry at the OPEN of the candle after the signal, exit after `h` candles
* fees on both legs, plus a slippage estimate per leg
* position sizing and a capital constraint
* a threshold on `direction_head` confidence, so low-conviction NEUTRAL calls
  are genuinely flat and do not pay fees for nothing
* walk-forward evaluation across several disjoint test periods, not one
* realised funding, if this were ever extended to perpetual futures

Report nothing until all of that exists and has been run on data the model never
saw.

## 14. Troubleshooting

**`ImportError: TensorFlow is not available`** - see section 0. Install it into
this environment or a `venv`; `requirements.txt` deliberately omits it.

**Training is far too slow** - lower `SUBSET_TRAIN`, lower `EPOCHS`, raise
`BATCH`, or lower `UNITS`. Measure one epoch first and extrapolate. Remember the
subset run is for iteration, not for reporting.

**Memory error while batching** - lower `BATCH`. The block is only ~43 MB, so
this is almost always the materialised batch, not the dataset.

**Killed during XGBoost** - set `RUN_XGBOOST = False`, or lower
`XGB_ESTIMATORS`, or raise `SUBSET_TRAIN`'s effect by reducing it.

**Numbers differ from a Colab run** - expected. Different device, different
kernel selection, different reduction order. Only a large, repeatable gap means
anything.

**`Could not find data/final/dataset_metadata.json`** - run
`python src/pipeline.py` and start Jupyter from the project root or from
`notebooks/`.

**Should I use this or notebook 04?** - 04 is the reference: full data, GPU,
and the configuration the project documents. This notebook is for iterating
quickly while editing features or targets locally.
